```python
# Data Visualization:
plt.hist(df["Age"], bins = 20)
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Histogram of Age")
plt.show()
```

Note:

. It's would seem that data with the Age gaps from 25 to 27 show up quite frequence in this dataset, compare to the rest of the data and other age.

```python
sns.countplot(data = df, x="Gender", hue = "Gender", palette = ["yellow", "teal"], legend = False)
plt.xlabel("Gender")
plt.ylabel("Count")
plt.title("Count of Genders")
plt.show()
```

Note:

. It's seem that the amount of data relate to those of the Male, in the Gender categories, seem to be higher than those in female.

```python
sns.boxplot(data = df, x="ExperienceInCurrentDomain")
plt.xlabel("Experience In Current Domain - in Years")
plt.title("Boxplot of Exeperience - Domain")
plt.show()
```

Note:

. It's seem that most people in the data have anywhere from 1 to 4 of experiences, and number higher than that must be a fews outlier, and it's seem that most people for this study is from 2 to 4 years of experience working in their field.

```python
plt.scatter(df["Age"], df["ExperienceInCurrentDomain"])
plt.xlabel("Age")
plt.ylabel("Experience in Current Domain")
plt.title("Experience Domain - Age Scatter Plot")
plt.show()
```
