```python
city_leave_counts = df.groupby("City")["LeaveOrNot"].value_counts().unstack()
city_leave_counts.plot(kind = "bar", stacked = True)
plt.xlabel("City")
plt.ylabel("Count")
plt.title("Bar Plot of City with Stacked Leave or Not")
plt.legend(title= "LeaveOrNot")
plt.show()
```
